```python
import requests
```

```python
from bs4 import BeautifulSoup
url = "https://ajuntament.barcelona.cat/calendarifestius/es/"

response = requests.get(url) # peticion http
print(response)
print(bool(response))
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
ul_list = soup.find_all("ul", attrs={"class": "dies"})
fest = ul_list[0]  # solo el primer <ul> 2025
```

```python
len(fest)
```

```python
fest.li.strong.text
```

```python
fest.li.span.text
```

```python
dia_festivo =[] # 1 de enero
tipo_festivo =[] # estatal
Name_festivo =[]
for li in fest.find_all("li"): 
    dia = li.strong # encontrar a 
    if dia: # condicio si a 
        dia_festivo.append(dia.get_text(strip=True)) # obtener y guardar
        
    tip = li.span
    if tip:
        tipo_festivo.append(tip.get_text(strip=True)) #obtener y guardar
        
    nom = li
    if nom:
        destino = "".join(nom.find_all(string=True, recursive=False)).strip() # obtener solo 
        #destino = destino.lstrip("- ").strip()    # eliminar "- " limpiar
        Name_festivo.append(destino)
        
print(dia_festivo)
print(tipo_festivo)
print(len(dia_festivo))
print(len(tipo_festivo))

print(len(Name_festivo))
print((Name_festivo))
```

```python
import pandas as pd
data = pd.DataFrame({'dia_festivo':dia_festivo, 'nombre_festivo':Name_festivo, 'tipo_festivo':tipo_festivo})
data
```

```python
meses = {
    "enero": 1,
    "febrero": 2,
    "marzo": 3,
    "abril": 4,
    "mayo": 5,
    "junio": 6,
    "julio": 7,
    "agosto": 8,
    "septiembre": 9,
    "octubre": 10,
    "noviembre": 11,
    "diciembre": 12
}
```

```python
fechas_convertidas = []

for texto in dia_festivo:       # ej. "1 de enero"
    partes = texto.split(" ")   # ["1", "de", "enero"]
    dia = int(partes[0])
    mes = meses[partes[2]]      # "enero" → 1
    fecha = pd.Timestamp(year=2025, month=mes, day=dia)
    fechas_convertidas.append(fecha)
```

```python
fechas_convertidas
```

```python
df_fest = pd.DataFrame({
    "Date_holiday": fechas_convertidas,
    "Name_holiday": Name_festivo,
    "Type_holiday": tipo_festivo
})
df_fest
```

```python
from datetime import datetime, timedelta
df_fest['holiday_eve'] = df_fest['Date_holiday'] - pd.Timedelta(days=1)
```
