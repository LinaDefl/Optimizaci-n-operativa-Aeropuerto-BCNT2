```python
df_fin["apertura_ci"] = df_fin["datetime_salida"] - pd.Timedelta(hours=2)
df_fin["cierre_ci"]   = df_fin["datetime_salida"] - pd.Timedelta(hours=1)
#df_fin["hora_apertura"] = df_fin["apertura"].dt.hour
# Redondear apertura y cierre a horas completas
# Pasaportes: Amplio (ejemplo: de 4h antes a 30min antes)
df_fin["apertura_pp"] = df_fin["datetime_salida"] - pd.Timedelta(hours=3)
df_fin["cierre_pp"] = df_fin["datetime_salida"] - pd.Timedelta(minutes=20)

# OBTENER HORA DE APERTURA Y CIERRE DEDONDEADO 
df_fin["apertura_hora"] = df_fin["apertura_ci"].dt.floor("h")
df_fin["cierre_hora"]   = df_fin["cierre_ci"].dt.floor("h")
df_fin
```

CHECK IN ->

```python
# --- DEFINIR PORCENTAJE DE PASAJEROS QUE PASAN POR CHECK-IN ---
# DEFINIR TRAMOS DE CHECK IN , define explícitamente quién entra al sistema

# Calcula cuántos pasajeros teóricos pasan por check-in
#Los distribuye minuto a minuto
#Normaliza para que la suma por vuelo sea correcta
# “De este vuelo, solo estos pasajeros pasan por check-in, 
#y lo hacen en este intervalo temporal”
PORC_CHECKIN = {
    "INT": 0.70,
    "EU":  0.60,
    "NAC": 0.30
}

# --- DEFINIR PESOS TEMPORALES POR TIPO DE VUELO (CHECK-IN) ---
PESOS_CI_INT = {1: 0.80, 2: 0.20}
PESOS_CI_EU_NAC = {1: 0.60, 2: 0.40}
```

```python
# EXPANDIR CADA VUELO A MINUTO
df_fin["rango_ci"] = df_fin.apply(
    lambda r: pd.date_range(r["apertura_ci"], r["cierre_ci"], freq="1min"),
    axis=1
)

# Expandir cada fila (vuelo) en múltiples filas por cada minuto del rango
df_exp_ci = (
    df_fin
    .explode("rango_ci")
    .rename(columns={"rango_ci": "timestamp_ci"})
    .reset_index(drop=True)
)
```

```python
# CREAR TRAMO 1 Y TRAMO 2
df_exp_ci["tramo_ci"] = np.where(
    df_exp_ci["timestamp_ci"] < df_exp_ci["apertura_ci"] + pd.Timedelta(hours=1),
    1,
    2
)

# contar minutos por vuelo y tramo 
minutos_por_tramo_ci = (
    df_exp_ci
    .groupby(["clave_unica_vuelo", "tramo_ci"])
    .size()
)
```

```python
# asignar peso por minuto segun tramo 

def peso_minuto_ci(row):
    clave = row["clave_unica_vuelo"]
    tramo = row["tramo_ci"]

    # Selección de pesos según tipo de vuelo
    if row["Type_f"] == "INT":
        pesos = PESOS_CI_INT
    else:  # EU + NAC
        pesos = PESOS_CI_EU_NAC

    minutos_tramo = minutos_por_tramo_ci.loc[(clave, tramo)]

    return pesos[tramo] / minutos_tramo
    
df_exp_ci["peso_ci_minuto"] = df_exp_ci.apply(peso_minuto_ci, axis=1)
```

```python
#Calcular PAX que realmente pasan por check-in
df_exp_ci["pax_checkin_teorico"] = (
    df_exp_ci["PAX"] *
    df_exp_ci["Type_f"].map(PORC_CHECKIN)
)
```

```python
# Calcular pax por minuto (sin normalizar aún)
df_exp_ci["pax_ci_minuto"] = (
    df_exp_ci["peso_ci_minuto"] *
    df_exp_ci["pax_checkin_teorico"]
)
# PASAJEROS POR MINUTO * PASAJEROS SOLO DE CADA TIPO TOTAL 
# OBTENEMOS TODOS NUESTROS PASAJEROS TOTALES FILTRADOS POR TURA
# MULTIPLICADO POR PASAJEROS POR MINUTO 
```

```python
suma_ci = df_exp_ci.groupby("clave_unica_vuelo")["pax_ci_minuto"].sum()

df_exp_ci["factor_normalizacion"] = df_exp_ci["clave_unica_vuelo"].map(suma_ci)

df_exp_ci["pax_ci_minuto"] = (
    df_exp_ci["pax_ci_minuto"] *
    df_exp_ci["pax_checkin_teorico"] /
    df_exp_ci["factor_normalizacion"]
)
```
