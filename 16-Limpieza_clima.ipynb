¿Qué columnas vas a obtener?

La API devuelve normalmente:

time → fecha y hora

temp → temperatura (°C)

dwpt → punto de rocío

rhum → humedad relativa (%)

prcp → precipitación (mm)

wdir → dirección del viento

wspd → velocidad del viento

wpgt → racha máxima

pres → presión hPa

tsun → insolación

y otros si están disponibles
