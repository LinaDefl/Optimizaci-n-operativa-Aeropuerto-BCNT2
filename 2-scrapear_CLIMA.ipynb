```python
from meteostat import Hourly, Point
from datetime import datetime, timedelta
import pandas as pd
import os


# Ubicación (latitud, longitud)
lat, lon = 41.1815, 2.0424 # Barcelona aeropuerto  
location = Point(lat, lon)
    
# Fechas
yesterday = datetime.today() 
    
# Datos diarios
now = datetime.today()
data_hourly = Hourly(location, now.replace(hour=0, minute=0, second=0), now).fetch()
print(data_hourly)
```

```python
data_hourly.drop(columns=['dwpt','rhum', 'snow'], inplace=True)
```

```python
import pandas as pd
# ASEGURAR QUE NO AÑADO LOS MISMOS DATOS (DUPLICADOS)
try: # intenta cargar el csv existente 
    df_csv = pd.read_csv("clima_bcn.csv", parse_dates=True, index_col=0) 
    #parse_dates=true convierte la primera columna (el índice del tiempo) en objetos datetime.
    # index-col=0 usa la primera columna como índice del dataframe 
    new_data = data_hourly[~data_hourly.index.isin(df_csv.index)] 
    #data_hourly es el DataFrame con los datos descargados de Meteostat.
    # ~ seleccionando solo las filas que aún no están en el CSV.
    # new_datacontiene únicamente las filas nuevas que faltan agregar.
    df_updated = pd.concat([df_csv, new_data]) # concatena datos antiguos + nuevos 
    df_updated.to_csv("clima_bcn.csv") # actualiza el csv 
    
except FileNotFoundError:
    data_hourly.to_csv("clima_bcn.csv") # si no existe crearlo 
```
